Lab manual brief: the goal of this lab is to measure the bandgap of GaAs, which is a commonly used semiconductor in high-frequency, high-power, and radiation-hardened electronics; solar cells; and some niche devices which I am not familiar with. The bandgap $\varepsilon$ is the energy difference between the highest-energy band filled by a valence electron in the ground state and the lowest-energy unfilled band. The two bands are called the 'valence band' and the 'conduction band' respectively, because valence electrons are normally trapped around their atom but when raised to the conduction band can jump to other atoms. The Fermi energy lies inside the bandgap. (Is it always in the middle?)

In this lab, we identify the bandgap energy by using the theory that photons having energy $h\nu$ greater than the bandgap can be absorbed by valence electrons (causing them to jump up in energy into the conduction band), whereas photons having energy lower than the bandgap will pass through undisturbed, resulting in a step change in the semiconductor's photon absorption capability from zero to something exactly at the bandgap. This is strictly true for a perfect crystal at absolute zero temperature, but the absorption capability step change will be broadened and shifted in frequency for crystals having defects and being measured at room temperature, as will be the case for our GaAs samples in this lab. ## How Sampling Works
The following list explains the function of Ocean Optics sampling components in the sampling process:
1. The user stores reference and dark measurements to correct for instrument response variables.
2. The light from the light source transmits through an optical fiber to the sample.
3. The light interacts with the sample.
4. Another optical fiber collects and transmits the result of the interaction to the spectrometer.
5. The spectrometer measures the amount of light and transforms the data collected by the spectrometer
into digital information.
6. The spectrometer passes the sample information to OOIBase32.
7. OOIBase32 compares the sample to the reference measurement and displays processed spectral
information.

Implication: the spectrometer compares reference measurements against the sample measurements. So long as the sample measurements are aligned in frequency with the reference measurements, all significant gains and losses are linear, and the measurement condition is the same apart from the presence of the sample, and the CCD response is proportional to the photon flux, then the transmission coefficient at each frequency point is simply the ratio of the CCD response with the sample over the response without.

Application Tips
If the signal you collect is saturating the spectrometer (intensity greater than 4000 counts), you can decrease the
light level on scale in scope mode by:

- Decreasing the integration time
- Attenuating the light going into the spectrometer
- Using a smaller diameter fiber
- Using a neutral density filter with the correct optical density

If the signal you collect has too little light, you can increase the light level on scale in scope mode by:

- Increasing the integration time
- Using a larger diameter fiber
- Removing any optical filters
